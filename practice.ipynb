1. Remove deuplicate users but preserve order 

    You receive user ID from logs 
        - Remove duplicates 
        - preserve first appearence order 
    logs=[101,102,101,103,102,104]

```python
logs=[101,102,101,103,102,104]
seen=set()
result=[]
for x in logs:
    if x not in seen:
        seen.add(x)
        result.append(x)
print(result)        
```

```python
ide-by-side comparison (this is the key table)
Function	 Returns	Needs list()?	Why
map()	        iterator	✅	To extract values
filter()	    iterator	✅	To extract values
zip()	        iterator	✅	To extract values
sorted()	        list	❌	Already a list
list.sort()	        None	❌	In-place
```

```python
Lambda vs def (decision rule)
Situation	Use
One-line logic	lambda
Used once	lambda
key= parameter	lambda
map / filter	lambda
Multiple lines	def
Reused function	def
Debugging needed	def

80% of lambdas are used for key= in sorted, min, max
```
