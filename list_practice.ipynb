```python
"""3. Find the element that appers more then half the time ( Majority element (> n/2 times ))

nums= [2,2,1,2,3,2,2]

expected ; 2 """

nums=[2,2,1,2,3,2,2]
value=len(nums)/2
high=None
hc=None

for i in range(0,len(nums)):
    count=0
    for j in nums:
        if nums[i]==j:
            count+=1
    if count>value:
      high=nums[i]
      hc=count
print(high) 
print(hc) 
```

```python
"""4. Closest number to Zero

nums=[-10,-5,2,4,-3] 

expected : -3 """

nums=[-10,-5,2,4,-3]
closest=None

for num in nums:
    if num<0:
        if closest is None or num>closest:
           closest=num
print(closest)               
```

```python
nums = [-10, -5, 2, 4, -3]

closest = nums[0]

for num in nums:
    if abs(num) < abs(closest):
        closest = num
    elif abs(num) == abs(closest) and num < closest:
        closest = num

print(closest)
```

```python
#bubble sort

nums=[5,1,4,2]

for i in range(0,len(nums)):
    for j in range(0,len(nums)-i-1):
        if nums[j]>nums[j+1]:
           nums[j],nums[j+1]=nums[j+1],nums[j]
print(nums)           
```
