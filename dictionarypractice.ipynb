```python
students={
    "name":"abc",
    "age":25,
    "city":"pune"
}

students["salary"]=50000 #add new key and value
print(students)
```

```python
#update the values

students["age"]=30
print(students)
```

```python
#remove values
del students["salary"]
print(students)
```

```python
for k in students:
    print(k)
    print(students[k])
```

```python
for v in students.values():
    print(v)
```

```python
for k,v in students.items():
    print(k,v)
```
