```python
# prime number check using functions

def prime(n):
    is_prime = True
    for i in range(2,n):
        if n%i == 0:
            is_prime = False
    return is_prime
prime(5)
```

```python
# factorial program using loop logic

def factorial(number):
    fact=1
    for i in range(number,1,-1):
        fact*=i
    return fact

factorial(5)    
```

```python
# factorial using recursion

def factorial(n):
    if n<1:
        return 1
    return n * factorial(n-1)

factorial(5)
```

```python
# sum of first N numbers using recursion

def sum_all(n):
    if n<1:
        return 0
    return n+sum_all(n-1)

sum_all(5)
```

```python
#palindrome using function     #num %10 === 234 remender (4)last digit we can extract      num//10 ==== 234 quotient (23) we can extract

def palindrome(number):
    original = number
    reverse = 0
    while(number>0):
         digit = number%10
         number = number//10
         reverse=reverse*10+digit
        
    if reverse == original:
         return True      
    else:
         return False

      
palindrome(121)
        
```

```python
# program to print sum of digits of a number

def sum_of_digits(number):
    sum=0
    while number>0:
        digit=number%10
        number=number//10
        sum+=digit
    return sum
print(sum_of_digits(45))
```

```python
#program to print product of digits of a number using recursion

def prod_digits(number):
    prod=1
    if number==0:
        return 1
    return number%10 * prod_digits(number//10)

print(prod_digits(85))
```

```python
#program to print sum of even digits of a number

def sum_of_even_digits(number):
    sum=0
    while number>0:
        digit = number%10
        number = number//10
        if digit%2 == 0:
            sum+=digit       
    return sum    
sum_of_even_digits(902)        

```

```python
# reverse a number

def rev(number):
    rev=0
    while number>0:
        rev=rev*10+(number%10)
        number=number//10
    return rev 
print(rev(455))   
```

```python
# program to find a largest digit in a number

def largest(number):
    largest = 0
    while number>0:
        digit = number%10
        number = number//10
        if digit>largest:
            largest=digit
    return largest              

print(largest(87))
```

```python
#program for reverse using recursion
def reverse_number(n, rev=0):
    if n == 0:
        return rev
    rev = rev * 10 + n % 10
    return reverse_number(n // 10, rev)

print(reverse_number(12345))   # Output: 54321
```

```python
#program to check if there repetaive digits

def repeat(number):
    digits=[]
    while number>0:
        digit=number%10
        if digit in digits:
            return True
        digits.append(digit)
        number=number//10
    return False
print(repeat(1233))    
```
