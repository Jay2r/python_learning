```python
def calculate_discount(total_price, discount_percentage):
    discount= (total_price*discount_percentage)/100
    final_price = total_price-discount
    print(f"the final price is {final_price}")
```

```python
calculate_discount(2000,50)
```

```python
import time 
start_time= time.time()
total=0
for i in range(1,1000001):
    total += i
end_time= time.time()
print("Sum is:", total)
print("Time taken to compute the sum:", end_time - start_time, "seconds")
```

```python
import time
def calculate_sum():
    total=0
    for i in range(1,1000001):
        total += i
    return total
start_time= time.time()
sum= calculate_sum()
end_time= time.time()
print("Sum is:", sum)
print("Time taken to compute the sum:", end_time - start_time, "seconds")   
```
