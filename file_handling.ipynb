10. Memory Comparison

 Method	         Memory Usage	    When to Use
 
 read()	            High	        Small files
readlines()	        High	        Need list of lines
for line in f	    Very low	    Large files

5. Memory Comparison
Method	            Memory Usage	Reason
read()	                High	    Loads entire file
readlines()	            Very High	Entire file + list
readline()	            Low     	One line at a time
for line in file	Lowest (best)	Uses iterator

```python
#write mode open file and overwrites it if no file present it creates new one f.write doesnt add  a new line by default
with open("input/demo.txt","w")as f:
    f.write("Hello\n")
    f.write("world\n")
```

```python
#here a means appended mode it does not override
#if file not exisits it creates one
with open("input/demo.txt","a")as f:
    f.write("this is appended")
```

```python
"""x- Exclusive create 

    - Create a new file 
    - If file already exists - Error """
with open("input/demo.txt","x") as f:
    f.write("this is appended")    

```

```python
"""r+ - Read+write (No truncate)

    - Read & write 
    - File must exist 
    - Does not delete content 
    - Writing happens from current pointer position."""

with open("input/test.txt","r+")as f:
     print(f.read(10))
     f.seek(9)  
     f.write("\nend")  
```

```python
with open("input/demo.txt","r+")as f:
    print(f.read(10))
    f.seek(6)
    f.write("c") #here it replaces "wo" i.e 6 and 7 har by ab
```

```python
"""w+ - Write +Read (Truncate)

    - Opens for read and write 
    - If file exists -> clear it first 
    - Create file if not exists"""

with open ("input/test.txt","w+") as f:
     f.write("Hello")
     f.seek(0)
     print(f.read())
```
