# Master Rule 

1. Start is included 
2. End is excluded 
3. Step decides the direction 
4. Wrong direction --> Empty List 
5. Python never throws error in slicing ( unlike indexing )

```python
a=[10,20,30,40,50,60,70,80,90]
print(a[-3:2:-2])
print(a[6:1:-3])
print(a[-1::-1])
print(a[::1]) #here end is length of a i.e 9 for both print statements
print(a[::-1])  # if only negative step is mentioned start is -1 and end is last element from right
```
