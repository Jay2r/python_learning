Shallow Copy :

Photocopy of a file that still points to the same attched documents

Deep Copy :

Completly new file with all documents duplicated 

Feature             Shallow copy                Deep Copy 

Outer Object        New                             New 
Inner Object        Shared                          New 
Speed               Faster                          Slower 
Nested              No                              Yes
Memory Usages       Less                            More 


When Deep copy is Required?
- Nested List 
- Dict if list 
- Json like data 
- Data pipeline ( Spark )


2. Why shallow copy is dangerous?

----> Because modifiting nested objects affects original 


3. Is slicing deep Copy?

----> No, Slicing is shallow copy 


4. Why deepcopy is slower ?

----> Because it recursively copies everything.

```python
cube=[[[1,2],[3,4]],[[5,6],[7,8]]]
flat=[x for x in cube for row in cube for x in row]#here cube is used twice
print(flat)
```

```python
cube=[[[1,2],[3,4]],[[5,6],[7,8]]]
flat=[x for layer in cube for row in layer for x in row]
print(flat)
```

```python
cube=[
    [[1,2],[3,4]],
    [[5,6],[7,8]]
]

flat=[]
for layer in cube:
    for row in layer:
        for x in row:
            flat.append(x)
print(flat)
```

```python
matrix=[
    [1,2,3],
    [4,5,6],
    [7,8,9]
]
from itertools import chain
flat=list(chain.from_iterable(matrix))
print(flat)
#takes one iterable matrix iterates over it and chains each inner iterable automattically
```

```python
flat=list(chain(*matrix))
print(flat)
#The * operator is argument unpacking chain([1,2,3],[4,5,6],[7,8,9])
```
