```python
#finally always runs whether error occurs or not.
#finallycannot be used alone try is mandatory
try:
    a=100
    b=0
    c=a/b
    print(c)
finally:
    print("execution completed")
```

```python
#user defined execption
class AgeError(Exception):  #here wecreatedexception pass means no code in that class just custom exception
    pass

try:
    age = 18

    if age < 18:
        raise AgeError("Age must be 18 or above")

    print("Eligible to vote")

except AgeError as e:
    print(e)

print("Program continues")

#we can use just raise exception but it willnot be specifiv so we create and use for the required things
```

```python
class InsufficientFunds(Exception):
    pass
```

```python
def withdraw(amount,balance):
    if balance < amount:
        raise InsufficientFunds("amount should be less than balance")
    
```

```python
try:
    withdraw(400,300)
except InsufficientFunds as e:
    print("error",e) 
print("continues")     
```
