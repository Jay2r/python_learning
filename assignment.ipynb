```python
# 1.	A number is entered. If the number is divisible by both 3 and 5, print "FizzBuzz". If only divisible by 3 print "Fizz", if only divisible by 5 print "Buzz", else print the number.

number = int(input("enter the number"))
if number%3 ==0 and number%5==0:
    print("FizzBuzz")
elif number%3==0:
    print("fizz")
elif number%5==0:
    print("buzz")   
else:
    print(number)         
```

```python
#Write a program to check whether a number is a strong number (sum of factorials of digits = number).

number = int(input("enter the number"))
strong=0
temp=number
while temp>0:
    digit=temp%10
    fact=1
    for i in range(digit,1,-1):
        fact*=i
    strong+=fact
    temp=temp//10
if (strong == number):
    print("it is a strong number")
else :
    print("it is not a strong number")


```

```python
#3.	Write a program to check whether a number is a Harshad (Niven) number.

number = int(input("enter a number"))
temp = number
sum = 0

while temp>0:
    digit=temp%10
    sum+=digit
    temp=temp//10
if number%sum == 0:
    print("it is a harshad number")
else:
    print("it is not a harshad number")        
```

```python
#4.	Write a program to find the reverse of a number and also check if it is a palindrome.

number = int(input("enter a number"))
reverse=0
temp=number
while temp>0:
    digit=temp%10
    reverse=reverse*10+digit
    temp=temp//10
print(reverse)
if(temp==reverse):
    print("it is a palindrome")  
else:
    print("not a palindrome")      
```

```python
#5.	Write a program to find the largest digit in a number.

number = int(input("enter the number"))
largest = 0
while number>0:
    digit = number%10
    number = number//10
    if digit>largest:
        largest=digit
print(largest)        
```

```python
#6.	Write a program to find the Sum of even digits in a number.

number = int(input("enter the number"))
sum = 0
while number>0:
    digit = number%10
    number = number//10
    if digit%2 == 0:
        sum+=digit
print(sum)    
```

```python
#Write a program to calculate income tax based on slabs.

income = int(input("enter the amount"))
tax=0

if income<=25000:
    tax=0
elif income<=50000:
    tax=(income-25000)*0.05
elif income<=100000:
    tax=(income-50000)*0.20+(25000*0.05)
else:
    tax=(income-100000)*0.30+(25000*0.05)+(50000*0.2)       
print("the tax is:",tax)  
```

```python
#12.	Write a program to find the count of prime digits in a number.

number=int(input("enter the number"))
count=0


while number>0:
    digit=number%10
    number=number//10
    is_prime=True
    for i in range(2,digit):
        if digit%i ==0:
            is_prime=False 
            break     
    if is_prime==True:
        count+=1        
print(count)              
```

```python
#13.	Write a program to generate all perfect numbers between 1 and N.

number = int(input("enter the range"))

if number<=0:
    print("enter greater than 1")
if number>=0:
    for i in range(1,number+1):
        sum=0
        for j in range(1,i):
            if i%j==0:
                sum+=j 
        if  sum==i:
            print(f"the number {i} is a perfect number")       
```
