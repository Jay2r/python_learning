```python
#16.	Write a function to check whether a number is automorphic (number ends with its square).

def auto(number):
    square=number*number
    num=number
    len=0
    while(num>0):
         num=num//10
         len+=1
    autom=square%(10**len)
    if autom==number:
        return True
    else:
        return False
d=auto(int(input("enter a no")))
print(d)    
```

```python
#18.	Write a function that returns True if a number contains only even digits.

def is_even(number):
    num=number
    count=0
    even_count=0
    while(num>0):
        digit=num%10
        num=num//10
        count+=1
        if digit%2 ==0:
            even_count+=1
    if count == even_count:
        return True
    else:
        return False
d=is_even(10)
print(d)        
        
```

```python
def is_even(number):
    num = abs(number)  # handle negative numbers
    
    if num == 0:
        return True
    
    while num > 0:
        digit = num % 10
        
        if digit % 2 != 0:
            return False   # stop early
        
        num = num // 10
    
    return True
```

```python
#19.	Write a function to find the difference between sum of odd digits and even digits.

def diff(number):
    num=number
    odd_sum=0
    even_sum=0
    while num>0:
        digit=num%10
        if digit%2 == 0:
            even_sum+=digit
        else:
            odd_sum+=digit
        num=num//10
    return odd_sum-even_sum

d=diff(1234)
print(d)            
```

```python
#20.	Write a function to calculate digital root of a number
def digi(number):
    num=number
    while(num>9):
        sum=0
        while(num>0):
            digit=num%10
            sum+=digit
            num=num//10
        num=sum    
    return num        
d=digi(532)
print(d)    
```

```python
#28.	Write a recursive function to reverse a number.

def rev(number):
    num=abs(number)
    if num==0:
        return 0
    remaining=num//10
    power=countdig(remaining)
    return (num%10)*(10 ** power)+rev(remaining)
d=rev(765)
print(d)
```

```python
#29.	Write a recursive function to check if a number is prime

def is_prime(number,i = 2):
    num=number
    if num <2:
        return False
    if num == i:
        return True
    if num%i==0:
        return False
    return is_prime(num, i+1)
```

```python
def gcd(a, b):
    a, b = abs(a), abs(b)

    if a == b:
        return a
    
    if a > b:
        return gcd(a - b, b)
    else:
        return gcd(a, b - a)
```
