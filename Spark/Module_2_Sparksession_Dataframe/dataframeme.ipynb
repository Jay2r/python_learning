```python
from pyspark.sql import SparkSession
spark=SparkSession.builder\
    .appName("MySparkSessionDemo")\
    .master("local[*]")\
    .config("spark.sql.shuffle.partitions","4")\
    .getOrCreate()

print("SparkSession created")
print("AppName",spark.sparkContext.appName)
print("master",spark.sparkContext.master)
print("Spark Version",spark.version)
```

1️⃣ from pyspark.sql import SparkSession
What is this?

pyspark → Python API for Apache Spark

sql → Module that supports structured data processing (DataFrames, SQL)

SparkSession → Entry point to work with Spark using DataFrame & SQL APIs

Why do we need SparkSession?

In older Spark versions:

SparkContext → for RDD

SQLContext → for DataFrames

HiveContext → for Hive support

From Spark 2.0+, SparkSession replaces all of them.
So now SparkSession is the main entry point to Spark.

```python
data=[
    (1,"A",30,"Mumbai"),
    (2,"B",31,""),
    (3,"C",32,"Pune"),
    (4,"D",33,"Bangalore"),
]
columns=["id","name","age","city"]
df=spark.createDataFrame(data,columns)
```

```python
df.show()
```

```python
df.printSchema()
```

```python
from pyspark.sql.types import *

schema = StructType([
    StructField("id",IntegerType(),True),
    StructField("name",StringType(),True),
    StructField("age",IntegerType(),True),
    StructField("city",StringType(),True)   
])
```

```python
df=spark.createDataFrame(data,schema)
df.show()
```
